## Tokenización
La tokenización es el proceso de dividir un texto en unidades más pequeñas llamadas tokens (palabras, frases, símbolos, etc.). Es convertir texto continuo en elementos individuales que la máquina puede procesar.

Ejemplo:

**Texto original:** "Hola, ¿cómo estás? ¡Muy bien!"

**Después de tokenización:**
["Hola", ",", "¿", "cómo", "estás", "?", "¡", "Muy", "bien", "!"]

### Tipos de tokenización:

- **Por palabras:** Separa en palabras individuales
- **Por oraciones:** Divide el texto en oraciones completas
- **Por caracteres:** Cada carácter es un token
- **Subword tokenization:** Divide palabras en partes (útil para palabras nuevas)

### Usos de la tokenización:

Preprocesamiento para análisis de texto, Análisis de frecuencia de palabras, Entrada para modelos de machine learning, Traducción automática, Análisis sintáctico, Extracción de características

## Stopwords

Las stopwords son palabras muy comunes que aparecen frecuentemente pero aportan poco valor semántico al análisis. Son palabras "vacías" de significado específico.

### Ejemplos en español:

- **Artículos:** el, la, los, las, un, una
- **Preposiciones:** de, en, a, por, para, con
- **Conjunciones:** y, o, pero, que, si
- **Pronombres:** yo, tú, él, ella, nosotros
- **Verbos auxiliares:** ser, estar, haber, tener

**Ejemplos en inglés:** a, an, the, and, or, but, in, on, at, to, for, of, with, by

¿Para qué sirven?
1. Filtrado y limpieza: Se eliminan antes del análisis para reducir ruido
Mejoran la eficiencia computacional
Reducen el tamaño del vocabulario

2. Análisis más preciso: Antes: "El gato está en la casa"
Después de remover stopwords: "gato casa"
Se enfoca en palabras con significado real

3. Aplicaciones: Búsqueda de información: Mejorar relevancia
Clasificación de textos: Reducir dimensionalidad
Análisis de sentimientos: Enfocarse en palabras emotivas
Modelado de temas: Identificar temas reales

**Consideraciones importantes:**

**No siempre se deben eliminar:** En análisis de sentimientos, palabras como "no", "muy" son cruciales
Dependen del contexto: "to be or not to be" perdería sentido sin stopwords
Varían por idioma: Cada idioma tiene sus propias stopwords

Las Stopwords son palabras con alta frecuencia de aparicion pero que no me aportan para la solución de la problemática. En este ejemplo tendríamos 8 tokens, "El", "gato" , "es", "negro", "y", "el", "perro", blanco" pero tenemos que procurar disminuir la dimensionalidad de la matriz y llegar a solo 4 tokens.

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

Mostrar los tokens que se encuentran dentro de la libreria nltk

```python
stop_words = set(stopwords.words('spanish'))
print(stop_words)
```

```python
texto2 = texto2.lower()
word_tokenized = word_tokenize(texto2)
print("Despues de usar  tokenizer:",word_tokenized)

filtered_sentence = [word for word in word_tokenized if not word in stop_words]

print("After removing stopwords:",filtered_sentence)
```

Comparado con el ejemplo 2 se puede ver que este modelo de tokenizer separa las comas en otro espacio, como un elemento, en lugar de dejarlo junto con la palabra previa.  

```python
def filter_text_stopwords(text):
    word_tokenized = word_tokenize(text.lower())
    return [w for w in word_tokenized if not w in stop_words]
```
