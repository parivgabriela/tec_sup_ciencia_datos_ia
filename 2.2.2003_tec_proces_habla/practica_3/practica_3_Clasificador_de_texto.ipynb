## Inicialización de los modelos

Creamos dos matrices de transicion, una para los archivos de Benedetti(clase 0), y el otro para Neruda(clase 1). Junto con la estructura(array) de las probabilidades.

### Clasificación de texto basado en un modelo de *Markov de primer orden*

Se crean dos conjuntos de matrices y vectores, uno para la clase 0 y otro para la clase 1.

**array_0 y array_1:** Son las futuras matrices de transición. array_0[i, j] almacenará la frecuencia con la que la palabra j aparece justo después de la palabra i en los textos de la clase 0.

**prob_et_0 y prob_et_1:** Son los futuros vectores de probabilidad inicial. prob_et_0[i] almacenará la frecuencia con la que la palabra i aparece al inicio de un texto de la clase 0.

**Punto clave: **Se inicializan con np.ones en lugar de ceros. Esta es una técnica de suavizado llamada Suavizado de Laplace (Laplace Smoothing). Sirve para evitar probabilidades de cero para secuencias de palabras que no aparecieron en el entrenamiento, lo que haría que el modelo fallara.

````python
len_indicepalabras = len(indicepalabras)
array_0 = np.ones((len_indicepalabras, len_indicepalabras))
prob_et_0 = np.ones(len_indicepalabras)

array_1 = np.ones((len_indicepalabras, len_indicepalabras))
prob_et_1 = np.ones(len_indicepalabras)
````

````python
array_0
````

````python
prob_et_0
````

````python
def compute_counts(text_as_int, A, pi):
    for tokens in text_as_int:
        last_idx = None
        for idx in tokens:
            # estamos en la primera palabra de la secuencia
            if last_idx is None:
                pi[idx] += 1
            else:
                A[last_idx, idx] += 1
            last_idx = idx

compute_counts([t for t, y in zip(train_text_int, Ytrain) if y == 0], array_0, prob_et_0)
compute_counts([t for t, y in zip(train_text_int, Ytrain) if y == 1], array_1, prob_et_1)
````

````python
# Normaliza A y pi para que sean matrices de probabilidad válidas

array_0 /= array_0.sum(axis=1, keepdims=True)
prob_et_0 /= prob_et_0.sum()

array_1 /= array_1.sum(axis=1, keepdims=True)
prob_et_1 /= prob_et_1.sum()
````

### La Clase Classifier

Es el "plano" o la "receta" para construir nuestro clasificador.

    __init__(self, logAs, logpis, logpriors): El Constructor

Se inicializa con los modelos ya entrenados.
- **logAs:** Las matrices de transición de log-probabilidades (```[log_array_0, log_array_1]```).

- **logpis:** Los vectores de log-probabilidades iniciales (```[log_prob_et_0, log_prob_et_1]```).

- **logpriors:** La log-probabilidad previa de cada clase (logp0, logp1), que representa qué tan común es cada clase en general.

    
    _compute_log_likelihood(self, input_, class_): El Motor de Probabilidad

Esta función es el núcleo del modelo. Calcula la log-verosimilitud (log-likelihood) de una secuencia de texto.

Responde a la pregunta: "Si asumimos que este texto pertenece a la clase_ X, ¿qué tan probable es esta secuencia de palabras según el modelo de lenguaje de esa clase?"

Lo hace sumando las log-probabilidades:

  - Empieza con la probabilidad de la primera palabra (logpi).

  - Luego, para cada palabra siguiente, suma la probabilidad de transición desde la palabra anterior (logA).

La suma se usa en lugar de la multiplicación porque se trabaja con logaritmos.

    predict(self, inputs): El Tomador de Decisiones

Este método toma una lista de textos (inputs) y predice la clase para cada uno.

Para cada texto, realiza el siguiente cálculo para cada clase posible (0 y 1):
        ```log(Posterior) = log(Likelihood) + log(Prior)```

Esto es una aplicación directa del teorema de Bayes en el espacio logarítmico:
        ```posteriors = [self._compute_log_likelihood(input_, c) + self.logpriors[c] ...] ```

Finalmente, np.argmax(posteriors) elige la clase que tuvo el valor de log-posterior más alto. Esa es la predicción final.

````python
# contrucción de un clasificador
class Classifier:
    def __init__(self, logAs, logpis, logpriors):
        self.logAs = logAs
        self.logpis = logpis
        self.logpriors = logpriors
        self.num_class = len(logpriors) # número de clases

    def _compute_log_likelihood(self, input_, class_):
        logA = self.logAs[class_]
        logpi = self.logpis[class_]

        last_idx = None
        logprob = 0
        for idx in input_:
            if last_idx is None:
                # Es el primer token en la secuencia
                logprob += logpi[idx]
            else:
                # Calcula la probabilidad de transición de la palabra anterior a la actual
                logprob += logA[last_idx, idx]

            # Actualiza last_idx para la próxima iteración
            last_idx = idx

        return logprob

    def predict(self, inputs):
        predictions = np.zeros(len(inputs))
        for i, input_ in enumerate(inputs):
            # Calcula los logaritmos de las probabilidades posteriores para cada clase
            posteriors = [self._compute_log_likelihood(input_, c) + self.logpriors[c] \
                          for c in range(self.num_class)]
            # Elige la clase con la mayor probabilidad posterior como la predicción
            pred = np.argmax(posteriors)
            predictions[i] = pred
        return predictions
````

Intancia la clase, se crea un objeto clasificador funcional

El código usa el clasificador para hacer predicciones tanto en el conjunto de entrenamiento (train_text_int) como en el de prueba (test_text_int).

Luego calcula la precisión (accuracy) comparando las predicciones (Ptrain, Ptest) con las etiquetas verdaderas (Ytrain, Ytest).

El "Test acc" (precisión de prueba) es el indicador más importante, ya que te dice qué tan bien generaliza tu modelo a datos que nunca ha visto.

````python
# Cada arreglo debe estar en orden ya que se asume que las clases indexan estas listas
clf = Classifier([log_array_0, log_array_1], [log_prob_et_0, log_prob_et_1], [logp0, logp1])
````

````python
Ptrain = clf.predict(train_text_int)
print(f"Train acc: {np.mean(Ptrain == Ytrain)}")
````

````python
Ptest = clf.predict(test_text_int)
print(f"Test acc: {np.mean(Ptest == Ytest)}")
````
