## ¿Qué es spaCy?

SpaCy es una biblioteca gratuita y de código abierto para el Procesamiento de Lenguaje Natural (PLN) avanzado en Python.

Si trabajas con grandes cantidades de texto, eventualmente querrás saber más sobre ese contenido. Por ejemplo: ¿De qué trata? ¿Qué significan las palabras en su contexto? ¿Quién hace qué a quién? ¿Qué empresas y productos se mencionan? ¿Qué textos son similares entre sí?

spaCy está diseñado específicamente para su uso en entornos de producción y te ayuda a crear aplicaciones que procesan y "entienden" grandes volúmenes de texto. Se puede usar para construir sistemas de extracción de información o de comprensión del lenguaje natural, o para preprocesar texto con fines de aprendizaje profundo (deep learning).

````python
import spacy
import pandas as pd
````

````python
nlp = spacy.load('es_core_news_sm')
````

````python
texto = """ Los estudiosos aman estudiar, es por eso que son felices estudiando.
            Su estudio es su felicidad.
            Por eso estudian felizmente sus estudios.
        """
documento = nlp(texto)
````

````python
print("Original\tLematizado")
for token in documento:
    if token.is_punct or token.is_stop or token.is_space:
        continue
    print(f"{token}  \t {token.lemma_}")
    print('  ')
````

### ¿Qué hace es_core_news_lg?

Este modelo permite a spaCy realizar tareas avanzadas de Procesamiento de Lenguaje Natural (PLN) en español, como:

- Tokenización: dividir el texto en palabras, signos de puntuación, etc.

- Lematización: obtener la forma base de las palabras.

- Etiquetado gramatical (POS tagging): identificar si una palabra es un verbo, sustantivo, adjetivo, etc.

- Análisis sintáctico (dependencias): entender las relaciones gramaticales entre palabras.

- Reconocimiento de entidades nombradas (NER): detectar nombres de personas, lugares, organizaciones, fechas, etc.

- Vectores semánticos: representar el significado de las palabras como vectores numéricos, lo que permite calcular similitud entre textos o palabras.
