## Limpieza de texto
Para evitar redundancias en el proceso de analizar el texto, se quitan los espacios, se convierte todo el texto a minúscula, para luego sacar los signos de puntuaciones que hay en el texto.

````python
def clean_text(text):
    s = text.rstrip().lower()
    return s.translate(str.maketrans('','',string.punctuation))
````

## Procesamiento de texto

Se utilizan diccionarios para almacenar los datos del texto, se almacenan las primeras palabras de cada texto leido, y luego cuenta, cuantas veces se repite esa palabra al inicio.

Primer orden almacena la primer palabra junto con las palabras que continuan, ejemplo 'el': ['perro', 'diario', 'hogar'] etc

segundo_orden, almacena las palabras del 'medio' la ```palabra anterior anterior``` y la palabra anterior, junto con la lista de todas las posibles palabras continuas, y en caso que lo amerite la palabra END en caso que termine

````python
pa_inicial = {}
primer_orden = {}
segundo_orden = {}
````

````python
def add2dict(dic, k, v):#Completa con datos el diccionario que le paso por parametros
    if k not in dic:
        dic[k] = []
    dic[k].append(v)
````

````python
indice = 0
with open(archivo, 'r', encoding='utf-8') as archivo:
    for line in archivo:
        tokens = clean_text(line).split()
        T = len(tokens)
        if indice < 15:
          print(f"Tokens: {tokens}\nTamaño de la fila: {T}")
          indice += 1
        for i in range(T):
            token = tokens[i]
            if i == 0:
                # condicional de la primer palabra
                # Si existe le suma uno, sino usa el 0 y le suma
                pa_inicial[token] = pa_inicial.get(token, 0.) + 1
            else:
                token_ant = tokens[i-1]
                if i == T - 1:
                    # condicional de la ultima posicion
                    add2dict(segundo_orden, (token_ant, token), 'END')
                if i == 1:
                    add2dict(primer_orden, token_ant, token)
                else:
                    token_ant_ant = tokens[i-2]
                    add2dict(segundo_orden, (token_ant_ant, token_ant), token)
````

## Normalizar: distribución de frecuencias

Actualizamos el valor de ```pa_inicial[clave]``` para que contenga la probabilidad de que esa palabra se de al principio.

Se repite la misma logica para los diccionarios que tienen iunformacion del texto.

````python
inicial_total = sum(pa_inicial.values())
print(inicial_total)
for clave, conteo in pa_inicial.items():
    pa_inicial[clave] = conteo / inicial_total
````

````python
# 'para': ['sobrevivirme', 'que', 'tus', 'que', 'tus', 'qué', 'mi', 'tu'],
def list2pdict(ts):
    d = {}  # Crear un diccionario vacío
    n = len(ts)  # Obtener la longitud de la lista de elementos

    # Ciclo para contar la ocurrencia de cada elemento en la lista
    for t in ts:
        d[t] = d.get(t, 0.) + 1

    # Ciclo para convertir los conteos en probabilidades relativas
    for t, c in d.items():
        d[t] = c / n

    return d  # Devolver el diccionario de probabilidades
````

````python
for t_1, ts in primer_orden.items():
    # replace list with dictionary of probabilities
    primer_orden[t_1] = list2pdict(ts)
````

````python
def palabra_ejemplo(d, imprimir):
    # Genera un número aleatorio en el rango (0, 1)
    p0 = np.random.random()
    if(imprimir == 1):
        print(f"p0: {p0}")

    # Inicializa una variable para realizar la suma acumulativa de probabilidades
    cumulative = 0
    if(imprimir == 1):
        print(f"prob acumulada: {cumulative}")

    # Ciclo que recorre cada clave (t) y su probabilidad (p) en el diccionario (d)
    for t, p in d.items():
        # Agrega la probabilidad actual al valor acumulativo
        cumulative += p
        if(imprimir == 1):
            print(f"item: {t}, Prob; {p}")
            print(f"prob acumulada: {cumulative}")

        # Comprueba si el número aleatorio es menor que la acumulación de probabilidades
        if p0 < cumulative:
            # Si se cumple la condición, devuelve la clave (t) seleccionada
            return t
````

## Generador de texto

````python
def generador(tamaño):
    for i in range(tamaño):
        oracion = []
        #Palabra Inicial
        pal0 = palabra_ejemplo(pa_inicial, 0)
        oracion.append(pal0)
        #Segunda Palabra
        pal1 = palabra_ejemplo(primer_orden[pal0], 0)
        oracion.append(pal1)

        # Segundo orden hasta el fin
        while True:
            pal2 = palabra_ejemplo(segundo_orden[(pal0, pal1)], 0)
            if pal2 == 'END':
                break
            oracion.append(pal2)
            pal0 = pal1
            pal1 = pal2
        print(' '.join(oracion))
````

````python
generador(5)
````
