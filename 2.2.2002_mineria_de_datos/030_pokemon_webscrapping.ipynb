Importando librerias necesarias

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Datos
Recolectar datos del sitio web de https://pokemondb.net/pokedex/all, donde se encuentran los datos de distintos pokemones, donde se ve una tabla con id, Name, Type, Total, HP, Attack, Sp.Atk, Sp.Def, Speed.

```python
url_pokemon = "https://pokemondb.net/pokedex/all"
```

```python
r = requests.get(url_pokemon)
```

Leer el contenido del codigo html

```python
r.content[:800]
#r.text
```

# Uso de la libreria Beautiful soup

BeautifulSoup es una biblioteca de Python que se utiliza para extraer información de archivos HTML y XML de forma sencilla y eficiente.

Traemos el html, y se imprime con el formato junto con los espacios

```python
soup = BeautifulSoup(r.content, "html.parser")
```

```python
# soup
```

Traemos solo el elemento de la tabla por el id 'pokedex'

```python
row_table = soup.find("table",{"id":'pokedex'})
type(row_table)
```

```python
elementos = row_table.findAll('tr') #['tr']
lista_elementos = list(elementos)
print(len(lista_elementos))
print(lista_elementos[0])
type(elementos)
```

```python
table = soup.find(id='pokedex')
elementos_row = table.find_all('tr') #['tr']

lista_elementos = list(elementos_row)
print("longitud: ", len(lista_elementos))
print("primer elemento", lista_elementos[0])
type(elementos_row)
```

  Ahora trabajaremos fila por fila para traer la información de la tabla

```python
data = []
nombres = []
tipos = []
totales = []
# hp = []
# ataques
# defense
stats = []
```

```python
index = 0
for row in soup.find_all('tr'):
    celdas = row.find_all('td')

    if len(celdas) < 10:
        continue  # Saltar filas incompletas

    nombre = celdas[1].find('a', class_='ent-name').text.strip()
    tipo_tags = celdas[2].find_all('a')
    tipo_textos = [tipo.text.strip() for tipo in tipo_tags]
    total = int(celdas[3].text.strip())
    stat_values = [int(td.text.strip()) for td in celdas[4:10]]

    nombres.append(nombre)
    tipos.append(tipo_textos)
    totales.append(total)
    stats.append(stat_values)

    if index < 10:
        print(f'Nombre: {nombre}, tipo: {tipo_textos}, total: {total}, HP: {stat_values[0]}, ataque: {stat_values[1]}, defensa: {stat_values[2]}, ')
    index += 1
print("Longitud: ", len(nombres))

```

```python
index = 0
for elemento in data:
    #datos_fila = elemento.get_text()
    if index < 5:
        print(elemento)
        break
    else:
        index += 1
```

```python
for i in range(len(nombres)//60):
    print(f"Nombre: {nombres[i]}")
    print(f"Tipos: {tipos[i]}")
    print(f"Total: {totales[i]}")
    print(f"Estadísticas: {stats[i]}")
    print("-" * 40)
```

```python
data_dict = {'nombre': nombres, 'tipos': tipos, 'totales': totales, 'estadisticas': stats}
```

## Creacion de un dataframe

Guardamos la informacion de las listas para poder operar con la libreria de pandas

```python
import pandas as pd

df_pokemon = pd.DataFrame(data_dict)
df_pokemon
```

## Exportar la informacion en CSV

```python
df_pokemon.to_csv('my_data.csv', index=False)
```
