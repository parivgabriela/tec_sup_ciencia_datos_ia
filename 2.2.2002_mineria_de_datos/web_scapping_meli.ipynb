# Web Scrapping
En este laboratorio se trabajó utilizando la tecnica de web scrapping, en la pagina de Mercadolibre, en la sección de Ofertas.

Lo que se desea obtener es recolectar la información y almacenarla en un archivo CSV.

Datos a recolectar: nombre del producto, precio de lista y su precio con descuento. Todo debe guardarse en un archivo .csv.

## Importación de librerías

- request
- BeautifulSoup
- Pandas

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd
````

```header``` es importante para simular un navegador y evitar bloqueos.

````python
url = "https://www.mercadolibre.com.ar/ofertas#c_id=/home/promotions-recommendations&c_uid=afc81145-8afd-4d44-a846-ce5bf340bd2e"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
}
````

get_html_from_url devuelve el objeto de soup con la informacion del link, en caso que falle la peticion devolvera ```None```

````python
def get_html_from_url(url, headers):
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        # Creamos un objeto BeautifulSoup para poder navegar por el HTML.
        soup = BeautifulSoup(response.content, "html.parser")
        return soup
    return None

main_content = get_html_from_url(url, headers)
# print(main_content)
````

````python
paginas = main_content.find_all("a", class_="andes-pagination__link")
print(paginas)
````

````python
tarjetas = main_content.find('div', class_="items-width-smart-groups")
print(tarjetas)
````

````python
def get_products(soup_element):
    lista_productos = []

    tarjetas = soup_element.find_all('div', class_='poly-card')
    print("tarjetas: ", len(list(tarjetas)))
    for tarjeta in soup_element.find_all('div', class_='poly-card'):
        articulo:str = tarjeta.find("a", class_="poly-component__title").get_text()
        precio_net: str = tarjeta.find("span", class_="andes-money-amount__fraction").get_text()
        precio_des: str = tarjeta.find("span", class_="andes-money-amount").get_text()
        lista_productos.append([articulo, precio_net, precio_des])
    return lista_productos
````

````python
lista_total = []

lista_total = get_products(main_content)
print(lista_total)
productos_df = pd.DataFrame(lista_total, columns=["articulos", "precio_neto", "precio_descuento"])
````
