<h2>Medidas de tendencia central en NumPy</h2>
<p>En estadística, las medidas de tendencia central son valores que intentan resumir un conjunto de datos en un solo número, representando el "centro" o el valor típico de ese conjunto. Las tres más importantes son la media, la mediana y la moda.</p>
<h3>Media o promedio</h3>
<p>La media es el valor promedio de un conjunto de datos. Se calcula sumando todos los valores y dividiendo por el número de valores.</p>


```python
array_nros = np.array([5,5,5,7,8,10,12,12,14,15,16,16,16,20,2,2,2,3,4,22,22])
media = np.mean(array_nros)
print(f"Promedio/Media: {media}")
print(f"Promedio/Media: {media:.2f}") # formato de 2 decimales con :.2f
```

### Mediana

La mediana es el valor que se encuentra justo en el medio de un conjunto de datos, una vez que ha sido ordenado de menor a mayor.

¿Cómo se calcula? Primero, ordena todos los valores de menor a mayor.

Encuentra el valor del centro. En caso de que la cantidad de elementos sea par, se realiza el proemedio de ambos valores.

```python
mediana = np.median(array_nros)
print(f"Mediana: {mediana}")
```

### Moda

Me devuelve un objeto de tipo ModeResult(array_con_el_primer_valor_que_encuentra, recuento_de_apariciones)
Sólo devuelve la primera ocurrencia en un vector ORDENADO si hay más valores, los ignora 

```python
moda = stats.mode(array_nros)
print("Moda:", moda)
```

## Medidas de posición

 Las medidas de posición te dicen dónde se encuentra un dato en relación con el resto del conjunto ordenado.

### Cuartiles

 Los cuartiles son el tipo más común de medida de posición. Son los tres valores que dividen un conjunto de datos ordenados en cuatro partes iguales, donde cada parte contiene el 25% de los datos.

CUARTILES --> numpy.quantile(datos, q)

```python
q1 = np.quantile(array_nros, 0.25)
q2 = mediana
q3 = np.quantile(array_nros, 0.75)

print(f"Q1 = {q1} - Q2 = {q2} - Q3 = {q3}")
```

### Percentil

Un percentil es un valor que divide un conjunto de datos ordenados en 100 partes iguales. El percentil k indica el valor por debajo del cual se encuentra un k por ciento de las observaciones.

```python
# la muestra en 100 => trabaja con porcentajes
# método percentile(arreglo, cuántos elementos tomar)
p60 = np.percentile(array_nros, 60)
p20 = np.percentile(array_nros, 20)
p50 = np.percentile(array_nros, 50)

print("Percentil 60:", p60) # el 60% de los datos ---> menor que q3, que es el 75%
print("Percentil 20:", p20) # el 20% de los datos ---> menor que q1, que es el 25%
print("Percentil 50:", p50) # el 50% de los datos ---> coincide ocnla mediana

```

## Medidas de variabilidad o dispersión

### Varianza

La varianza es una medida de dispersión que indica qué tan dispersos están los datos en relación con la media. Matemáticamente, la varianza se define como el promedio de los cuadrados de las diferencias entre cada valor y la media del conjunto de datos.
En NumPy, la varianza se puede calcular utilizando el método `var()`.

```python
varianza = np.var(array_nros)
print(f"Varianza: {varianza}")
print(f"Varianza: {varianza:.2f}")
```

### Desvío Estándar

Es la raíz cuadrada de la varianza

```python
desvio_std = np.std(array_nros)
print(f"Desvío stardard: {desvio_std:.2f}")
print(desvio_std**2)
```

<h3>Arreglos vacíos</h3>
<p>A veces se crean arreglos vacíos, para luego ser rellenados.</p>
<p>Se inicializan con unos, ceros, valores espaciados uniformemente, valores constantes o valores aleatorios. </p>

```python
# arreglo de unos --> float
unos = np.ones(3)
unos2 = np.ones((4,5)) # doble paréntesis
unos3 = np.ones((4,5), dtype=np.int64) # enteros

print("Vector 1x3:\n", unos)
print("\nMatriz 4x5:\n", unos2)
print("\nMatriz 4x5:(int)\n", unos3)
```

```python
# arreglo de ceros 
unos = np.zeros(3)
unos2 = np.zeros((4,5)) # doble paréntesis
unos3 = np.zeros((4,5), dtype=np.int64) # enteros

print(unos)
print()
print(unos2)
print()
print(unos3)
```

### Random
arreglo de nros aleatorios  --> método random

```python
# SUBMÉTODO random() --> nro aleatorio entre 0 y 1
a1 = np.random.random(3)
print(a1)
print()
a2 = np.random.random((3,5))
print(a2)
print()
# SUBMÉTODO randint()
a3 = np.random.randint(1,6,5) # random.randint(mínimo, máximo(-1), cantidad)
print(a3)
print()
a4 = np.random.randint(1,6,(4,5))
print(a4)
```

Para crear arreglos con valores espaciados uniformemente podemos utilizar np.arange() y np.linspace()


```python
# np.arange(): deseo crear un arreglo con los valores comprendidos entre 0 y 100 a paso 5
esp1 = np.arange(0,101,5)
print(esp1)
print()
# np.linspace(): deseo crear 5 valores que se encuentren entre 0 y 2
# esp2 = np.arange(0,2.1,0.5) 
esp2 = np.linspace(0,2,5) # linspace(inicio, tope(entra), cantidad)
print(esp2)
# np.linspace(min, max, n) me divide el rango en min-max en n partes y llena el arreglo con los valores que me da la división
 
```

<h3>Repaso de acceso a los elementos y slicing</h3>
Para obtener fragmentos de nuestros sets de datos.

```python
a = np.array([[1,0,3],[4,3,5],[6,10,-1]])
print(a)
```

Acceder al 10 (recuerdo que las posiciones empiezan a contarse desde 0)

```python
print(a[2,1])
print()
print(a[1:,1:]) # desde la fila 1, y desde la columna 1
```

```python
# recordemos que si estamos en un array de una sola dimensión, ponemos únicamente 
# el índice que corresponde al elemento que buscamos y listo!
a2 = np.array([25, 36, 10, 85, 56])
print(f"elemento 2 {a2[2]}")
print(f"elemento hasta el ':2'{a2[:2]}")
print(f"elemento ultimo(-1) {a2[-1]}")

```

```python
# SLICING TOTAL -> Para obtener todos los elementos
a2 = np.array([25, 36, 10, 85, 56])
print(a2[:])
print(a2[::2]) # se queda con los elementos de 2 en 2 posiciones
```
