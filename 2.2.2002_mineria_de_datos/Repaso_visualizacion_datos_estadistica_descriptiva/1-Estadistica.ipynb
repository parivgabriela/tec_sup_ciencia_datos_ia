## Variables discretas y continuas

<center><img src="img/discretas_continuas.PNG"></center>


<center><h4>Que una variable sea discreta, no significa que sea número entero</h4><center>
<p>La longitud es un dato continuo, pero la escala en la regla es una variable discreta, porque (aunque tiene 'decimales') no nos es posible ponerle un valor exacto a todos los espacios entre una 'marca' y la siguiente</p>

<h2>Tipos de métricas descriptivas</h2>

<h3>1. Medidas de tendencia central</h3>
<p>Las medidas de tendencia central son medidas estadísticas que pretenden resumir en un solo valor a un conjunto de valores. Representan un centro en torno al cual se encuentra ubicado el conjunto de los datos. Las medidas de tendencia central más utilizadas son: <b>media, mediana y moda.</b> Nos indican cómo están distribuidos los datos.</p>
<center><img src="img/tend_central.png">
<em><small>Imagen: Allende H y AhumadaS, ILI-280</small></em></center>

<h3>2. Medidas de dispersión</h3>
<p>Las medidas de dispersión entregan información sobre la variación de la variable. Pretenden resumir en un solo valor la dispersión que tiene un conjunto de datos, teniendo en cuenta los valores obtenidos en el estudio de tendencia central. Las medidas de dispersión más utilizadas son: <b>rango de variación, varianza, desviación estándar y coeficiente de variación.</b> Indican cómo se alejan los datos respecto de la media aritmética y sirven como indicadores de la variabilidad de los datos.</p>
<center><img src="img/dispersion.png">
<em><small>Imagen: Wikipedia</small></em></center>



<em><small>Ilustraciones de https://storyset.com/</em></small>

<h2>Medidas de tendencia central</h2>
<p>Son una manera de resumir información de toda una población y lo más básico en cuanto a análisis de datos.</p>


<h3>Moda</h3>
<p>Representa el dato que más se repite en una distribución de datos. Puede ser unimodal, bimodal o polimodal.</p>
<p><b>Ejemplo:</b> Buscamos qué cantidad de personas de cada edad hay en el aula.
<p>Si no hay gran cantidad de datos, pueden representarse en una <b>tabla de frecuencias.</b></p>
<p>Si el rango de datos es muy amplio, conviene utilizar un <b>diagrama de frecuencias.</b></p>
<center><img src="img/frecuencias.png"></center>
<p>La moda no aplica para datos numéricos continuos</p>

<h3>Media</h3>
<p>Es lo que conocemos comúnmente como <b>promedio</b></p>
<ul>
<li>Ventaja: Permite extraer información valiosa de los datos rápidamente.</li>
<li>Desventaja: Hay valores extremos que pueden distorsionar la información.</li>
</ul>
<p>Si tomamos como ejemplo el caso de las edades de las personas del aula, el promedio da 20,97. Este valor está más cercano a 21, que no representa a la mayoría de la muestra. 
<em><small>Foto de harith irfan wahid en Unsplash</small></em></center>
<p>Supongamos que nos piden calcular el promedio de salario de los 10 gerentes que están en este momento en la cafetería, y que por política de la empresa, todos los gerentes cobran $250.000. El cálculo del promedio sería sencillo: <b><code>10 x $250.000 / 10 = $250.000</code></b></p>
<p>¿Qué sucedería si de repente ingresa a la cafetería el accionista mayoritario de la empresa, que obtiene ganancias de la empresa por $3.500.000 al mes?</p>
<p>Ahora el cálculo sería: <b><code>[(10 x $250.000) + $3.500.000] / 11 = $545.454,54,</code></b> valor bastante lejano a la realidad. Por eso se dice que la media está <b>sesgada</b> por los valores atípicos.</p>


<h3>Mediana</h3>
Aquí se trata de obtener el valor que se encuentra exactamente en el medio del set de datos, <b>una vez ordenados.</b>
<ul>
<li>Cuando los datos son impares: es el dato que está justo en el centro.</li>
<li>Cuando los datos son pares: se hace un promedio de los dos datos que están en el centro.</li></ul>
<p>En el ejemplo de las edades, la mediana da 20:</p>
<center><img src="img/mediana.png"></center>
<p>En el ejemplo de los sueldos de los gerentes, la mediana da $250.000:</p>
<center><img src="img/mediana2.png"></center>
<p>Como podemos apreciar, la mediana nos da una idea más precisa, porque no se ve afectada por los valores de los extremos, y no interesa si se alejan demasiado de la mediana. Sería la opción más adecuada cuando nos encontramos con valores atípicos. Ésta es la diferencia fundamental entre la media y la mediana.</p>

<h1>Medidas de dispersión</h1>
<h3>Descripción por datos</h3>
<p>Podemos describir nuestras muestras extrayendo datos como los cuartiles, máximos y mínimos, mediana, promedio, desviación standard. </p>


```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('data/cars_depurado.csv')
df
```

```python
# Rango
rango = df['price_usd'].max() - df['price_usd'].min()
rango
```

## Cuartiles

```python
mediana = df['price_usd'].median() # Q2 --> 50% de los datos quantile(q=0.5)

# método quantile(q=0-1) según el valor del parámetro q, sabremos en qué parte de la subdivisión
# de la distribución de datos estamos
Q1 = df['price_usd'].quantile(q=0.25)
Q3 = df['price_usd'].quantile(q=0.75)

Q2 = df['price_usd'].quantile(q=0.50)
# valores mínimos y máximo
min_val = df['price_usd'].quantile(q=0) # df['price_usd'].min()
max_val = df['price_usd'].quantile(q=1) # Q4 -> df['price_usd'].max()

print(f'Min = {min_val} - Q1 = {Q1} - Q2 = {mediana} - Q3 = {Q3} - Q4 = {max_val}')
print(f"Q2 cuartil: {Q2}")
```

```python
# RANGO INTERCUARTIL -> IQR
iqr = Q3 - Q1
iqr
```

## Varianza

Tomamos el cuadrado de la distancia entre la media y un punto cualquiera del set de datos
Hacemos el promedio de todas esas distancias.
Tomamos los cuadrados para evitar los valores negativos. Así obtenemos la VARIANZA

```python
print("Desviación standard: ", round(df['price_usd'].std(),2))
print("Varianza: ", round(df['price_usd'].var(),2))
```

<h3>Límites para detección de valores anómalos</h3>
<p>Datos entre <code><mark>Q<sub>1</sub> - 1,5 x IQR</mark></code> y <code><mark>Q<sub>3</sub> + 1,5 x IQR</mark></code>

```python
# RECORDAMOS: este método se utiliza SI TUVIÉRAMOS una distribución normal
minlimit = Q1 - 1.5 * iqr
maxlimit = Q3 + 1.5 * iqr
print(f"Limite inferior: {minlimit}, Limite superior: {maxlimit}")
```

EJERCICIO: calcular promedio, mediana, min, max, q1, q2, q3, iqr, desviación standard de precios en un df con datos de una marca de coches a elección. Graficar histograma, caja y scatterplot(precio/año)

```python
df_volks = df[df['manufacturer_name'] == 'Volkswagen']
```

```python
#promedio 
promedio_df_volks = df_volks['price_usd'].mean()
#mediana y q2
mediana_df_volks = df_volks['price_usd'].median()
#min
min_df_volks = df_volks['price_usd'].min()
max_df_volks = df_volks['price_usd'].max()
q1_df_volks = df_volks['price_usd'].quantile(q=0.25)
q3_df_volks = df_volks['price_usd'].quantile(q=0.75)
iqr_df_volks = q3_df_volks - q1_df_volks
#desviación standard
std_df_volks = df_volks['price_usd'].std()

print(f"Promedio: {promedio_df_volks}")
print(f"Mediana: {mediana_df_volks}")
print(f"Min: {min_df_volks}")
print(f"Max: {max_df_volks}")
print(f"Q1: {q1_df_volks}")
print(f"Q2: {mediana_df_volks}")
print(f"Q3: {q3_df_volks}")
print(f"IQR: {iqr_df_volks}")
print(f"Desviación standard: {std_df_volks}")
```
