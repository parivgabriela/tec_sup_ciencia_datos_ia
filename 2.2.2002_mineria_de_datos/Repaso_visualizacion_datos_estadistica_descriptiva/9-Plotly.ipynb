# Plotly

## Introducción

Plotly Express es una biblioteca de visualización de datos de alto nivel, construida sobre la biblioteca Plotly. Su principal objetivo es permitir la creación de gráficos estadísticos atractivos e interactivos con muy pocas líneas de código.
Posee una amplia variedad de recursos y gráficos únicos, estadísticos, geográficos, científicos, econométricos y tridimensionales.

Al igual que Matplotlib y Seaborn, se pueden combinar los valores **numéricos** y **categóricos** y se combina para trabajar con pandas de forma simple.
[plotly documentos](https://plotly.com/python/)

### ¿En qué casos se usa?
Se utiliza principalmente para el análisis exploratorio de datos (EDA). Es ideal cuando necesitas entender rápidamente la distribución, la relación y la composición de tus datos. Algunos ejemplos comunes de gráficos que puedes crear son:

**Gráficos de dispersión** para ver la relación entre dos variables.

**Gráficos de líneas** para observar tendencias a lo largo del tiempo.

**Histogramas y diagramas de caja** para entender la distribución de una variable.

**Gráficos de barras** para comparar categorías.

**Mapas para visualizar** datos geográficos.

**Gráficos circulares y de torta** para mostrar proporciones.

## Dataset

````
df_tips = px.data.tips()
ds(df_tips)
````

### Barras

````
px.bar(df_tips, x='day', y="total_bill", color="smoker",barmode="group", title='Valor de la cuenta x día / fumador/no fumador')
````

### Histogramas

````
px.histogram(df_tips, x='total_bill', title='Valor total de la cuenta')
````

````
# Valores estadisticos
def calcular_medidas_tendencia_central(df, column):
    promedio = df[column].mean()
    mediana = df[column].median()
    moda = df[column].mode().iloc[0]
    return promedio, mediana, moda
````

````
promedio, mediana, moda = calcular_medidas_tendencia_central(df_tips, 'total_bill')
print(f"El Promedio es: {promedio:.2f}")
print(f"La Mediana es: {mediana:.2f}")
print(f"La Moda es: {moda:.2f}")

# 2. Crear el histograma base
fig = px.histogram(df_tips, x='total_bill', title='Distribución del Valor Total de la Cuenta')

# 3. Añadir las líneas verticales para la mediana y la moda
fig.add_vline(x=mediana,
              line_width=3,
              line_dash="dash", # Estilo de línea
              line_color="red", # Color
              annotation_text=f'Mediana: {mediana:.2f}', # Etiqueta
              annotation_position="top right")

fig.add_vline(x=moda,
              line_width=3,
              line_dash="dot", # Estilo de línea diferente
              line_color="green", # Color diferente
              annotation_text=f'Moda: {moda:.2f}',
              annotation_position="top left")

fig.show()
````

````
# Mostramos los primeros 5 registros del dataframe para entender su estructura
print("Vista previa del dataset de propinas:")
ds(df_tips.head())

# Creamos un gráfico de dispersión interactivo
# Puedes pasar el cursor sobre los puntos para ver los datos exactos.
fig_scatter = px.scatter(df_tips,
                         x='total_bill',
                         y='tip',
                         color='sex', # Color por género
                         facet_col='day', # Crea una columna por cada día
                         title='Relación Cuenta Total vs. Propina por Día y Género',
                         labels={'total_bill': 'Cuenta Total ($)', 'tip': 'Propina ($)', 'sex': 'Género'})

# Creamos un boxplot para ver la distribución de la cuenta total por día
fig_box = px.box(df_tips,
                 x='day',
                 y='total_bill',
                 color='smoker', # Diferenciamos entre fumadores y no fumadores
                 title='Distribución de la Cuenta Total por Día (Fumadores vs. No Fumadores)',
                 labels={'day': 'Día de la Semana', 'total_bill': 'Cuenta Total ($)', 'smoker': 'Fumador'})


# Usamos ds() para mostrar ambos gráficos en la misma celda
print("\n--- Gráficos Interactivos ---")
ds(fig_scatter)
ds(fig_box)
````
